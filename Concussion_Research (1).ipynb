**RATE RATIO TEST: Is a kickoff more dangerous per play?**

```python
from scipy.stats import norm

def rate_ratio(cases_a, plays_a, cases_b, plays_b):
    rr = (cases_a / plays_a) / (cases_b / plays_b)
    se = np.sqrt(1 / cases_a + 1 / cases_b)
    low = rr * np.exp(-1.96 * se)
    high = rr * np.exp(1.96 * se)
    p_value = 2 * norm.sf(abs(np.log(rr) / se))
    return rr, low, high, p_value

risk_lookup = risk_graph.set_index("PlayType")

rate_ratio_rows = []

for period in ["1996_2001", "2002_2007"]:
    for compare_to in ["Passing", "Rushing", "Punt"]:
        rr, low, high, p_value = rate_ratio(
            risk_lookup.loc["Kickoff", "Concussions_" + period],
            risk_lookup.loc["Kickoff", "Plays_" + period],
            risk_lookup.loc[compare_to, "Concussions_" + period],
            risk_lookup.loc[compare_to, "Plays_" + period]
        )
        rate_ratio_rows.append({
            "Period": period.replace("_", "-"),
            "Comparison": "Kickoff vs " + compare_to,
            "Rate Ratio": round(rr, 2),
            "95% CI": f"{low:.2f} to {high:.2f}",
            "P-value": p_value
        })

rate_ratios = pd.DataFrame(rate_ratio_rows)

rate_ratios
```

```python
period_rows = []

for play in risk_lookup.index:
    rr, low, high, p_value = rate_ratio(
        risk_lookup.loc[play, "Concussions_2002_2007"],
        risk_lookup.loc[play, "Plays_2002_2007"],
        risk_lookup.loc[play, "Concussions_1996_2001"],
        risk_lookup.loc[play, "Plays_1996_2001"]
    )
    period_rows.append({
        "PlayType": play,
        "Rate Ratio (2002-2007 vs 1996-2001)": round(rr, 2),
        "95% CI": f"{low:.2f} to {high:.2f}",
        "P-value": round(p_value, 3),
        "Significant": p_value < 0.05
    })

period_change = pd.DataFrame(period_rows)

period_change
```

**ROBUSTNESS CHECK: Chi-square tests without sparse cells**

```python
print("Original play type test: cells with expected count < 5:",
      f"{(expected < 5).mean() * 100:.1f}%")
print("Original position test: cells with expected count < 5:",
      f"{(expected_pos < 5).mean() * 100:.1f}%")

# Play type: merge rare plays into "Other", drop "Unknown"
play_merged = play_graph.copy()
play_merged["PlayType"] = play_merged["PlayType"].replace({
    "Change": "Other",
    "Field goal attempt": "Other",
    "Point after touchdown": "Other"
})
play_merged = play_merged[play_merged["PlayType"] != "Unknown"]
play_merged = play_merged.groupby("PlayType", as_index=False)[
    ["Count_1996_2001", "Count_2002_2007"]
].sum()

chi2_r, p_r, dof_r, expected_r = chi2_contingency(
    play_merged[["Count_1996_2001", "Count_2002_2007"]]
)

print("\nPlay type (merged, no Unknown)")
print("Chi-square statistic:", chi2_r)
print("Degrees of freedom:", dof_r)
print("P-value:", p_r)
print("Cells with expected count < 5:", f"{(expected_r < 5).mean() * 100:.1f}%")

# Position: merge kickers, punter, and holder into "Specialists", drop "Unknown"
pos_merged = positions_graph.copy()
pos_merged["Position"] = pos_merged["Position"].replace({
    "Punter": "Specialists",
    "Kicker, FGA": "Specialists",
    "Kicker, PAT": "Specialists",
    "Kicker, KO": "Specialists",
    "Holder": "Specialists"
})
pos_merged = pos_merged[pos_merged["Position"] != "Unknown"]
pos_merged = pos_merged.groupby("Position", as_index=False)[
    ["Count_1996_2001", "Count_2002_2007"]
].sum()

chi2_pr, p_pr, dof_pr, expected_pr = chi2_contingency(
    pos_merged[["Count_1996_2001", "Count_2002_2007"]]
)

print("\nPosition (merged, no Unknown)")
print("Chi-square statistic:", chi2_pr)
print("Degrees of freedom:", dof_pr)
print("P-value:", p_pr)
print("Cells with expected count < 5:", f"{(expected_pr < 5).mean() * 100:.1f}%")
```

```python
def adjusted_residuals(table, labels):
    observed = table.values
    _, _, _, exp = chi2_contingency(observed)
    total = observed.sum()
    row_share = observed.sum(axis=1, keepdims=True) / total
    col_share = observed.sum(axis=0) / total
    resid = (observed - exp) / np.sqrt(exp * (1 - row_share) * (1 - col_share))
    return pd.DataFrame({
        "Category": labels.values,
        "Adjusted Residual (2002-2007)": resid[:, 1].round(2),
        "Significant": np.abs(resid[:, 1]) > 1.96
    }).sort_values("Adjusted Residual (2002-2007)", ascending=False)

print("Play type (merged, no Unknown)")
display(adjusted_residuals(
    play_merged[["Count_1996_2001", "Count_2002_2007"]],
    play_merged["PlayType"]
))

print("Position (merged, no Unknown)")
display(adjusted_residuals(
    pos_merged[["Count_1996_2001", "Count_2002_2007"]],
    pos_merged["Position"]
))
```

# **Risk Assessment Tool GUI**

```python
import ipywidgets as widgets
from IPython.display import display, Markdown
```

```python
position_scores = positions_graph.copy()

position_scores["RiskScore"] = (
    position_scores["Count_2002_2007"] /
    position_scores["Count_2002_2007"].max() * 100
).round(1)
```

```python
# Use the exposure-adjusted rate (risk per 1,000 plays) instead of raw
# concussion counts, so the score reflects true per-play danger rather than
# how often a play type just happens to occur. This only covers the 4 play
# types play_risk_data.csv has rate data for (Kickoff, Punt, Rushing, Passing).
play_scores = risk_graph.copy()

play_scores["RiskScore"] = (
    play_scores["Risk_per_1000_2002_2007"] /
    play_scores["Risk_per_1000_2002_2007"].max() * 100
).round(1)
```

```python
activity_scores = activity_graph.copy()

activity_scores["RiskScore"] = (
    activity_scores["Count_2002_2007"] /
    activity_scores["Count_2002_2007"].max() * 100
).round(1)
```

```python
position_dropdown = widgets.Dropdown(
    options=position_scores["Position"].tolist(),
    description="Position:"
)

play_dropdown = widgets.Dropdown(
    options=play_scores["PlayType"].tolist(),
    description="Play:"
)

activity_dropdown = widgets.Dropdown(
    options=activity_scores["Activity"].tolist(),
    description="Activity:"
)

button = widgets.Button(
    description="Calculate Risk",
    button_style="success"
)

output = widgets.Output()

display(position_dropdown)
display(play_dropdown)
display(activity_dropdown)
display(button)
display(output)
```

```python
def calculate_risk(button):

    output.clear_output()

    with output:

        try:

            plt.close('all')  # avoid piling up unclosed figures across clicks

            position = position_dropdown.value
            play = play_dropdown.value
            activity = activity_dropdown.value

            # Look up the selected rows
            position_row = position_scores[
                position_scores["Position"] == position
            ]

            play_row = play_scores[
                play_scores["PlayType"] == play
            ]

            activity_row = activity_scores[
                activity_scores["Activity"] == activity
            ]

            # Extract risk scores
            position_score = position_row["RiskScore"].iloc[0]
            play_score = play_row["RiskScore"].iloc[0]
            activity_score = activity_row["RiskScore"].iloc[0]

            # Weighted overall score
            overall_score = round(
                position_score * 0.40 +
                play_score * 0.35 +
                activity_score * 0.25,
                1
            )

            # -----------------------------------
# Estimate uncertainty
# -----------------------------------

            position_error = abs(
                position_row["Percent_1996_2001"].iloc[0] -
                position_row["Percent_2002_2007"].iloc[0]
            ) / 2

            play_error = abs(
                play_row["Percent_1996_2001"].iloc[0] -
                play_row["Percent_2002_2007"].iloc[0]
            ) / 2

            activity_error = abs(
                activity_row["Pct_1996_2001"].iloc[0] -
                activity_row["Pct_2002_2007"].iloc[0]
            ) / 2

            overall_uncertainty = round(
                position_error * 0.40 +
                play_error * 0.35 +
                activity_error * 0.25,
                1
            )

            # Determine risk level
            if overall_score >= 75:
                risk_level = "🔴 HIGH"
            elif overall_score >= 50:
                risk_level = "🟡 MODERATE"
            else:
                risk_level = "🟢 LOW"

            from IPython.display import display, HTML

            display(HTML("""
<h2 style='color:#1f77b4; text-align:center;'>
🏈 Football Concussion Risk Assessment Tool
</h2>
<hr>
"""))

            risk_bar = widgets.IntProgress(
                value=int(overall_score),
                min=0,
                max=100,
                description='Risk:',
                bar_style='danger' if overall_score >= 75 else
                          'warning' if overall_score >= 50 else
                          'success'
            )

            display(risk_bar)

            print(f"\nPosition : {position}")
            print(f"Play Type: {play}")
            print(f"Activity : {activity}")

            print("\n-------------------------------")

            print(f"Position Score : {position_score:.1f}/100")
            print(f"Play Score     : {play_score:.1f}/100")
            print(f"Activity Score : {activity_score:.1f}/100")

            print("\n-------------------------------")

            display(HTML(f"""
<h1 style='text-align:center; color:#d62728; margin-bottom:0px;'>
{overall_score:.1f}/100
</h1>

<h3 style='text-align:center; color:gray; margin-top:5px;'>
Estimated Uncertainty: ±{overall_uncertainty:.1f}
</h3>

<p style='text-align:center; font-size:15px;'>
Historical variability between the two study periods
(1996–2001 and 2002–2007)
</p>
"""))

            color = "red" if overall_score >= 75 else \
        "orange" if overall_score >= 50 else \
        "green"

            display(HTML(f"""
<h3 style='color:{color};'>
Risk Level: {risk_level}
</h3>
"""))

            print("📊 Why this score?")

            if position_score >= 75:
                print(f"• {position} has historically experienced relatively high concussion totals.")

            if play_score >= 75:
                print(f"• {play} has historically been associated with a high number of concussions.")

            if activity_score >= 75:
                print(f"• {activity} frequently appeared in historical concussion reports.")

            print("🛡️ Recommendations")

            if overall_score >= 75:
                print("• Increase concussion monitoring.")
                print("• Reinforce proper tackling technique.")
                print("• Continue improving player safety protocols.")

            elif overall_score >= 50:
                print("• Monitor high-risk situations carefully.")
                print("• Reinforce safe playing techniques.")

            else:
                print("• Continue following current concussion prevention practices.")

            plt.figure(figsize=(6,4))

            plt.bar(
    ["Position", "Play Type", "Activity"],
    [position_score, play_score, activity_score],
    yerr=[position_error, play_error, activity_error],
    capsize=8
)

            plt.ylim(0,100)
            plt.ylabel("Risk Score")
            plt.title("Risk Score Breakdown")

            plt.show()

        except Exception as e:
            print("ERROR:")
            print(e)
```

```python
button.on_click(calculate_risk)
```
